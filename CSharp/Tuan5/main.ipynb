 Yêu cầu :

    - Khi nhập vào ComboBox, chương trình dò tìm từ khớp với ký tự gần nhất (auto-complete).

# Bài 2: Xử lý ListBox số nguyên

 Main idea : Nhập số vào TextBox → thêm vào ListBox. Các button thực hiện: tính tổng, xóa đầu/cuối, xóa phần tử chọn, tăng mỗi phần tử lên 2, thay bằng bình phương, chọn số chẵn/lẻ.

 Yêu cầu :

    - Button "Nhập": thêm số từ TextBox vào ListBox.
    - Button "Tổng": tính tổng các phần tử, hiển thị MessageBox.
    - Button "Xóa đầu và cuối": xóa phần tử đầu và cuối.
    - Button "Xóa phần tử đang chọn": xóa các item đang được select.
    - Button "Tăng mỗi phần tử lên 2": cộng 2 vào mọi phần tử.
    - Button "Thay bằng bình phương": bình phương hóa mọi phần tử.
    - Button "Chọn số chẵn": highlight các số chẵn.
    - Button "Chọn số lẻ": highlight các số lẻ.

 Output : 

    Input ListBox: 30, 21, 35, 43, 1, 3, 20, 9, 4, 80

    - Tổng các phần tử: 246
    - Sau khi xóa đầu và cuối: 21, 35, 43, 1, 3, 20, 9, 4
    - Sau khi tăng mỗi phần tử lên 2: 23, 37, 45, 3, 5, 22, 11, 6
    - Sau khi thay bằng bình phương: 529, 1369, 2025, 9, 25, 484, 121, 36
    - Chọn số chẵn: highlight 529 (lẻ), 1369 (lẻ), 2025 (lẻ), 9 (lẻ), 25 (lẻ), 484 (chẵn), 121 (lẻ), 36 (chẵn)
      → highlight: 484, 36

# Bài 4: Đồng hồ đếm ngược

 Main idea : Dùng `Timer` với Interval = 1000ms (1 giây) để đếm ngược từ 30:00. Button "Bắt đầu" toggle giữa Start/Stop.

 Yêu cầu :

    - Hiển thị thời gian dạng MM:SS.
    - Nhấn "Bắt đầu" → bắt đầu đếm ngược.
    - Nhấn lần nữa → dừng.
    - Khi hết giờ → hiện MessageBox "Hết giờ!".

 Output : 

    Ban đầu: 30:00
    Nhấn "Bắt đầu" → button đổi thành "Dừng"
    Sau 1 giây: 29:59
    Sau 2 giây: 29:58
    ...
    Nhấn "Dừng" → button đổi thành "Bắt đầu", timer dừng
    Khi về 00:00 → MessageBox: "Hết giờ!"
